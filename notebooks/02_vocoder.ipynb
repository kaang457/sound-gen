# 02 · (Opsiyonel) Kendi vocoder'ımız: WaveFlow (mel → ses)
> **Rota v3'te zorunlu değil:** ana vocoder hazır NVIDIA BigVGAN-v2. Bu notebook kendi WaveFlow'umuzu aynı mel
> tanımıyla (44,1 kHz, 128 bant, hop 512) eğitir; sonra BigVGAN'la birebir kıyaslanabilir ya da yerine takılabilir.
**`MODE = "vocoder"` (varsayılan):** WaveFlow yalnız 128 bantlık, 44,1 kHz mel'i görür (BigVGAN ile aynı tanım); kategori almaz.
Sesin NE olduğuna mel üretici karar verir (03), vocoder onu sese çevirir. Bu yüzden paketteki bütün sesler
(etiketsizler dahil) kullanılır ve eğitimde hiç görmediği türden (ör. sentetik sci-fi) mel'leri de çalabilir.
`MODE = "category"`: eski poc yolu (kategori + mel). `FIXES` base_v1 sağlık raporundaki düzeltmeleri açar.
**Paket:** 01'in 48 kHz paketi (`pack48_*`), eğitimde 44,1 kHz'e çevrilir. 22 kHz paketle başlatılırsa eğitim hemen net bir hatayla durur.
**Compute bütçesi:** `MAX_HOURS` dolunca eğitim kendiliğinden durur; adım sayısı süreden hesaplanır.
Kesilirse hücre 3'ü tekrar çalıştır: Drive'daki `last.pt`'den devam eder.
**Bitince tek bir sade çıktı:** GitHub `results/<RUN>/` → `README.md`, `showcase.html` (bütün sesler tek sayfada),
`health.html` (sağlık raporu), `model.pt` (üretim ağırlıkları) ve `sketch_bank.pt`. Checkpoint'ler ve ara örnekler Drive'da kalır.

```python
#@title 0 · Kurulum: repo + Drive (her yeni oturumda bir kez çalıştır)
import os, subprocess, sys
REPO = "/content/sound-gen"
try:
    from google.colab import userdata
    _tok = userdata.get("GITHUB_TOKEN")
except Exception:
    _tok = None
_url = f"https://x-access-token:{_tok}@github.com/kaang457/sound-gen" if _tok else "https://github.com/kaang457/sound-gen"
if not os.path.exists(REPO):
    subprocess.run(["git", "clone", "-q", _url, REPO], check=True)
else:
    subprocess.run(["git", "-C", REPO, "pull", "-q", "--ff-only"], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--no-deps", "-e", REPO], check=True)  # Colab's torch/numpy untouched
sys.path.insert(0, REPO)
for _m in [m for m in sys.modules if m == "soundgen" or m.startswith("soundgen.")]:
    del sys.modules[_m]          # re-import: picks up fixes pulled above without a restart
from soundgen import colab
P = colab.setup()          # Drive: MyDrive/sound-gen/{pack, runs, custom}
```

```python
#@title 1 · Ayarlar
RUN = "vocoder_v1"       #@param {type:"string"}
MODE = "vocoder"         #@param ["vocoder", "category"]
# vocoder: yalnız mel (rota v2) · category: kategori + mel (poc_v1 yolu)
PACK = "pack48_poc"      #@param {type:"string"}
# pack48_poc: 01'in poc_max verisi (lisans filtresiz) · pack48_full: yalnız ticari-temiz veri
UPSAMPLER = "interp"     #@param ["interp", "transposed"]
# interp: H1 düzeltmesi (vızıltı testi); transposed: makaledeki / base_v1
STEPS = 0                #@param {type:"integer"}
# 0 = süreden hesapla (MAX_HOURS dolana kadar; öğrenme oranı takvimi buna göre kurulur)
MAX_HOURS = 5.0          #@param {type:"number"}
FIXES = True             #@param {type:"boolean"}
# base_v1 sağlık raporundaki düzeltmeler: taslak normalizasyonu, ActNorm, dropout 0, grad_clip 25
PRESET = "auto"          #@param ["auto", "a100", "l4", "t4"]
import torch
gpu = torch.cuda.get_device_name(0) if torch.cuda.is_available() else ""
if PRESET == "auto":
    PRESET = "a100" if ("A100" in gpu or "H100" in gpu) else "l4" if "L4" in gpu else "t4"
PRESETS = {  # bf16 on A100/L4/H100; T4 runs fp32 -> smaller model
    "a100": dict(model=dict(res_channels=128), batch_size=8),
    "l4":   dict(model=dict(res_channels=96), batch_size=8),
    "t4":   dict(model=dict(res_channels=64), batch_size=4),
}
for p_ in PRESETS.values():
    p_["model"]["upsampler"] = UPSAMPLER
    if FIXES:
        p_["model"].update(cond_norm=True, actnorm=True, dropout=0.0)
        p_["grad_clip"] = 25.0
    p_["vocoder"] = MODE == "vocoder"
# 44,1 kHz: vocoder 0,35 s kesit (≈ poc'nin 16384 örneği), category 0,7 s (2 kat). A100 değerleri poc_v1 ölçümünden tahmin
SEC_PER_STEP = {"a100": 0.5, "l4": 1.1, "t4": 1.8} if MODE == "vocoder" else {"a100": 0.95, "l4": 2.1, "t4": 3.4}
if STEPS <= 0:
    STEPS = int(MAX_HOURS * 3600 / SEC_PER_STEP[PRESET] * 0.95)
print(gpu, "->", PRESET, PRESETS[PRESET], "| adım:", STEPS)
```

```python
#@title 2 · Kuru deneme: gerçek veriyle ~10 dk + yayın (~10 dk). Büyük eğitimden önce BİR KEZ çalıştır
# Kontrol eder: Drive'da yer, eğitim gerçek veride çalışıyor mu (kayıp düşüyor, NaN yok), sağlık raporu,
# vitrin ve GitHub'a gönderim. Sonuç results/<RUN>_dry/ altına gider (sonra silinebilir).
DRY_RUN = True          #@param {type:"boolean"}
DRY_MINUTES = 10        #@param {type:"number"}
import shutil, math, pandas as pd
free_gb = shutil.disk_usage(str(P.drive)).free / 1e9
print(f"Drive boş alan: {free_gb:.1f} GB", "✅" if free_gb > 6 else "⚠️ checkpoint'ler için en az ~6 GB gerekli")
if DRY_RUN:
    from soundgen.train import TrainConfig, train
    pack = colab.local_copy(P.pack.parent / PACK, f"/content/{PACK}")
    dry = P.runs / f"{RUN}_dry"
    n = max(300, int(DRY_MINUTES * 60 / SEC_PER_STEP[PRESET]))
    train(TrainConfig(packs=[str(pack)], out_dir=str(dry), steps=n, max_minutes=DRY_MINUTES, warmup=min(1000, n // 5),
                      eval_every=max(100, n // 4), sample_every=n, keep_steps=[], **PRESETS[PRESET]))
    log = pd.read_csv(dry / "log.csv")
    tr, va = log.train_nll.dropna(), log.val_nll.dropna()
    k = max(1, len(tr) // 4)
    checks = {
        "kayıp düşüyor (train)": (tr.iloc[-k:].mean() < tr.iloc[:k].mean()) if len(tr) >= 4 else "yetersiz veri",
        "kayıp düşüyor (val)": (va.iloc[-1] < va.iloc[0]) if len(va) >= 2 else "yetersiz veri",
        "NaN / sonsuz yok": all(math.isfinite(v) for v in list(tr) + list(va)),
    }
    files = colab.publish(dry, [pack], refine=None)
    checks["yayın: 5 dosya"] = len(files) >= 5
    for name, v in checks.items():
        print("⚠️" if isinstance(v, str) else "✅" if v else "❌", name, f"({v})" if isinstance(v, str) else "")
    print()
    print("Hepsi ✅ ise hücre 3 ile büyük eğitimi başlat. GitHub'da results/" + f"{RUN}_dry/ klasörünü kontrol et.")
```

```python
#@title 3 · Eğit ve yayınla (kaldığı yerden devam eder; bitince sade sonuç GitHub results/<RUN>/'a)
from soundgen.train import TrainConfig, train
pack = colab.local_copy(P.pack.parent / PACK, f"/content/{PACK}")   # Drive -> yerel disk (hızlı okuma)
cfg = TrainConfig(packs=[str(pack)], out_dir=str(P.runs / RUN), steps=STEPS, max_minutes=MAX_HOURS * 60,
                  eval_every=1000, sample_every=5000, keep_steps=[10000, 20000, 30000], **PRESETS[PRESET])
train(cfg)
# Bitince: vitrin + sağlık raporu + model -> Drive runs/<RUN>/publish ve GitHub results/<RUN>/ (~10 dk)
colab.publish(P.runs / RUN, [pack], refine=None)
```

```python
#@title 4 · Kayıp eğrisi
import pandas as pd, matplotlib.pyplot as plt
log = pd.read_csv(P.runs / RUN / "log.csv")
fig, ax = plt.subplots(figsize=(9, 3.5))
t, v = log.dropna(subset=["train_nll"]), log.dropna(subset=["val_nll"])
ax.plot(t.step, t.train_nll.rolling(10, min_periods=1).mean(), label="train NLL")
ax.plot(v.step, v.val_nll, "o-", label="val NLL (EMA)")
ax.set_xlabel("adım"); ax.set_ylabel("nats / örnek"); ax.legend(); plt.show()
```
